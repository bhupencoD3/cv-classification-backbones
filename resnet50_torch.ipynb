###### Imports

```python
import torch
import torchvision.transforms as transforms
from torchvision import models
from PIL import Image
```

###### Load the pretrained model

```python
model = models.resnet50(pretrained=True)
model.eval()
```

###### Image Processing

```python
transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485,0.465,0.406],std=[0.220,0.224,0.225])
])
```

###### Loading Image for Prediction

```python
img_path = 'indian-bustard.jpg'
img_raw = Image.open(img_path)
image = transform(img_raw).unsqueeze(0)
```

###### Model Inferencing

```python
with torch.no_grad():
    outputs = model(image)

_,indices = torch.topk(outputs,k=3)
indices = indices.squeeze().tolist()
```

###### Final Results

```python
from torchvision.models import ResNet50_Weights
from matplotlib import pyplot as plt
labels = ResNet50_Weights.DEFAULT.meta['categories']

predictions = [labels[idx] for idx in indices]
for i,label in enumerate(predictions):
    print(f'{i+1}: {label}')

plt.imshow(img_raw)
plt.title(f'Top Prediction: {predictions[0]}')
plt.axis('off')
plt.show()
```
